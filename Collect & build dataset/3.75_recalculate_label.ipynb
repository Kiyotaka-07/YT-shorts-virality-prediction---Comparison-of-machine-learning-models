```python
import pandas as pd
import numpy as np

df = pd.read_csv('dataset_final_1.csv')
```

```python
df = df.drop_duplicates(subset=["video_id"])

df["viral_score"] = (
    0.6 * df["views_log"] +
    0.3 * df["likes_log"] +
    0.1 * df["comments_log"]
)

threshold = df["viral_score"].quantile(0.8)

print("Global threshold:", threshold)
```

```python
df["viral"] = (
    df["viral_score"] >= threshold
).astype(int)
```

```python
print(df["viral"].value_counts())
```

```python
df.to_csv(
    "FINAL_dataset.csv",
    index=False
)
```
